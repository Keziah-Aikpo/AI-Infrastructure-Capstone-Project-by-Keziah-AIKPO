```python
countries = [
    "United States",
    "Germany",
    "France",
    "Norway",
    "Poland",
    "Japan"
]

for country in countries:
    print(country)
    print("Carbon:", country in carbon["Country"].values)
    print("Productivity:", country in productivity["Country"].values)
    print("Electricity:", country in electricity["Country"].values)
    print("Renewables:", country in renewables["Country"].values)
    print("Tax:", country in tax["Country"].values)
    print("-"*30)
```

```python
print("Carbon:", carbon["Year"].min(), "-", carbon["Year"].max())
print("Productivity:", productivity["Year"].min(), "-", productivity["Year"].max())
print("Electricity:", electricity["Year"].min(), "-", electricity["Year"].max())
print("Renewables:", renewables["Year"].min(), "-", renewables["Year"].max())
print("Tax:", tax["Year"].min(), "-", tax["Year"].max())
```

```python
len(set(productivity["Country"]))
len(set(carbon["Country"]))
len(set(electricity["Country"]))
len(set(renewables["Country"]))
len(set(tax["Country"]))
```

```python
common_countries = (
    set(productivity["Country"])
    & set(carbon["Country"])
    & set(electricity["Country"])
    & set(renewables["Country"])
    & set(tax["Country"])
)

print("Common countries:", len(common_countries))
```

```python
master = productivity.merge(
    carbon,
    on=["Country", "Year"],
    how="inner"
)

master = master.merge(
    electricity,
    on=["Country", "Year"],
    how="inner"
)

master = master.merge(
    renewables,
    on=["Country", "Year"],
    how="inner"
)

master = master.merge(
    tax,
    on=["Country", "Year"],
    how="inner"
)
```
